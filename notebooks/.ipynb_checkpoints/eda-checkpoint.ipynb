# Chicago City

```python
with open("../chicago-city.geojson", "r", encoding="utf-8") as file:
    geojson = json.load(file)

features = geojson["features"]

print("GeoJSON Type:", geojson["type"])
print("Number of Features:", len(features))
```

```python
# Part 1

records = []

for i, feature in enumerate(features):
    geometry = feature.get("geometry")
    properties = feature.get("properties") or {}

    records.append({
        "feature_id": i,
        "geometry_type": geometry.get("type") if geometry else None,
        "properties_count": len(properties),
        "has_geometry": geometry is not None
    })

df = pl.DataFrame(records)

print("Dataset Shape:")
print(df.shape)

print("\nDataset Schema:")
df.glimpse()

print("\nMissing Values:")
print(df.null_count())

print("\nDescriptive Statistics:")
print(df.describe())
```

### Part 1: Dataset Profiling: Chicago City GeoJSON

The Chicago City GeoJSON dataset contains one geographic feature representing Chicago's city boundary. The geometry type is MultiPolygon, which means the boundary can consist of multiple polygons.

I converted the feature information into a Polars DataFrame with 1 row and 4 columns for profiling. The columns include feature ID, geometry type, property count, and whether geometry is available.

The profiling results showed no missing values in these four fields. The feature contains valid geometry information, but no additional properties.

Since this dataset represents a geographic boundary rather than individual observations, descriptive statistics such as mean and standard deviation are not very meaningful. The coordinates will be more useful for mapping and geographic comparisons.

This dataset can help identify Chicago's geographic boundaries when analyzing crime and camera locations.

```python
# Part 2

print("Missing Geometry:")
print(sum(f.get("geometry") is None for f in features))

print("\nDuplicate Features:")
print(len(features) - len({
    json.dumps(f, sort_keys=True) for f in features
}))

rings = []
coordinates = []

for feature in features:
    geometry = feature.get("geometry")

    if geometry is None:
        continue

    if geometry["type"] == "MultiPolygon":
        polygons = geometry["coordinates"]
    elif geometry["type"] == "Polygon":
        polygons = [geometry["coordinates"]]
    else:
        print("Unsupported geometry:", geometry["type"])
        continue

    for polygon in polygons:
        for ring in polygon:
            rings.append(ring)

            for point in ring:
                coordinates.append(point)

print("\nTotal Coordinate Points:")
print(len(coordinates))

print("\nTotal Polygon Rings:")
print(len(rings))

invalid_coordinates = 0

for point in coordinates:
    longitude, latitude = point[:2]

    if not (-180 <= longitude <= 180 and
            -90 <= latitude <= 90):
        invalid_coordinates += 1

print("\nInvalid Coordinates:")
print(invalid_coordinates)

unclosed_rings = 0
short_rings = 0

for ring in rings:
    if len(ring) < 4:
        short_rings += 1

    if ring[0] != ring[-1]:
        unclosed_rings += 1

print("\nUnclosed Polygon Rings:")
print(unclosed_rings)

print("\nRings With Fewer Than Four Points:")
print(short_rings)

consecutive_duplicates = 0

for ring in rings:
    for i in range(1, len(ring)):
        if ring[i] == ring[i - 1]:
            consecutive_duplicates += 1

print("\nConsecutive Duplicate Points:")
print(consecutive_duplicates)
```

### Part 2: Anomaly Detection:Chicago City GeoJSON

I checked the Chicago City GeoJSON dataset for missing geometry, duplicate features, invalid coordinates, and problems with polygon boundaries.

The dataset contains 12,731 coordinate points across 5 polygon rings. There were no missing geometries, duplicate features, invalid coordinates, or unclosed polygon rings. All polygon rings also contained at least four points.

However, I found 156 consecutive duplicate coordinate points. These may be redundant vertices in the geographic boundary, but they do not necessarily make the geometry invalid.

I decided to leave these points unchanged and flag them rather than remove them, since modifying the coordinates could affect the original geographic boundary. No features or coordinate points were removed during this analysis.

```python
# Part 3: Geographic Visualization

plt.figure(figsize=(9, 10))

for ring in rings:
    longitudes = [point[0] for point in ring]
    latitudes = [point[1] for point in ring]

    plt.plot(
        longitudes,
        latitudes,
        linewidth=0.8
    )

plt.title("Chicago City Geographic Boundary")
plt.xlabel("Longitude")
plt.ylabel("Latitude")

plt.axis("equal")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

### Part 3: Geographic Visualization: Chicago City

The plot shows Chicago's geographic boundary using latitude and longitude coordinates. The smaller shapes represent additional polygon areas in the dataset.

This map helps us understand Chicago's geographic coverage and can be useful for analyzing crime and Flock camera locations.

## Relationship Between Ward Income and Crime Counts

```python
import polars as pl
# Load Chicago crime dataset
crimes_df = pl.read_csv(
    "../Crimes_-_2026_20260928.csv",
    infer_schema_length=10000,
    ignore_errors=True
)

print("Crime Dataset Shape:")
print(crimes_df.shape)

print("\nCrime Dataset Columns:")
print(crimes_df.columns)

print("\nFirst 5 Rows:")
print(crimes_df.head())
```

```python
# Count crimes in Chicago's 50 wards
crime_counts = (
    crimes_df
    .filter(
        pl.col("Ward").is_between(1, 50)
    )
    .group_by("Ward")
    .agg(
        pl.len().alias("crime_count")
    )
    .sort("Ward")
)

print(crime_counts)
```

```python
# Load Chicago ward income dataset
income_df = pl.read_csv("../chicago-ward-income.csv")

# Check column names before joining
print("Income Dataset Columns:")
print(income_df.columns)

print("\nCrime Dataset Columns:")
print(crime_counts.columns)
```

```python
# Join ward income with crime counts
combined_df = income_df.join(
    crime_counts,
    left_on="ward",
    right_on="Ward",
    how="inner"
)

# Keep the columns needed for our analysis
combined_df = combined_df.select(
    "ward",
    "median_household_income_est",
    "crime_count"
)

print("Combined Dataset Shape:")
print(combined_df.shape)

print("\nFirst 5 Rows:")
print(combined_df.head())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(9, 6))

plt.scatter(
    combined_df["median_household_income_est"],
    combined_df["crime_count"],
    alpha=0.7
)

plt.xlabel("Median Household Income ($)")
plt.ylabel("Reported Crime Count")
plt.title("Chicago Ward Income vs. Reported Crimes (2026)")

plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

### Part 4: Observations

The scatter plot compares median household income with reported crime counts across Chicago's 50 wards.

The Pearson correlation coefficient is -0.34, indicating a weak-to-moderate negative relationship. This suggests that wards with higher median household incomes tend to have fewer reported crimes, although there are several exceptions.

The relationship is not strong, and income alone cannot explain differences in crime counts.

These findings do not establish causation. Crime counts are not adjusted for population, and the income estimates are from 2023 while the crime records are from 2026.


## Part 5: So What?

Our analysis found a weak-to-moderate negative correlation (-0.34) between median household income and reported crime counts across Chicago's 50 wards.

This finding is relevant to our research because differences in neighborhood income may be associated with differences in reported crime. When examining whether Flock cameras help reduce crime, we should consider other factors, such as income, rather than assuming that differences in crime counts are caused by camera presence.

However, our analysis does not directly measure the effectiveness of Flock cameras. Further analysis is needed to examine camera locations, crime rates, and changes in crime over time.

## Part 6: What's Next?

Our analysis found a weak-to-moderate negative relationship between median household income and reported crime counts across Chicago's 50 wards.

However, there are several limitations. The income data is from 2023, while the crime data is from part of 2026. We also used total reported crime counts rather than crime rates adjusted for population.

For future analysis, we could:

1. Compare Flock camera locations with crime patterns across Chicago wards.
2. Calculate crime rates using population data to make comparisons between wards more meaningful.
3. Examine crime trends before and after Flock camera installation, if installation dates are available.
4. Investigate whether neighborhood income is associated with camera placement.

Based on these findings, we could refine our research question to:

"How is Flock camera presence associated with reported crime patterns across Chicago neighborhoods, after considering differences in neighborhood income and population?"

This revised question would help us investigate the relationship between cameras and crime while considering other factors that may influence the results.

# Research relationship plots

This notebook compares traffic-camera violations and speed-camera records over the years. The aim is to see whether the presence of flock cameras have any impact on speed and red light violations.

## Traffic Violations

Compare the monthly average reported violations per exported record for red-light and speed cameras. This is not a Flock-camera outcome measure.

```python
red_light = pl.read_csv("../red_light_100000.csv", schema_overrides={"VIOLATIONS": pl.String}).select(
    pl.col("VIOLATION DATE")
    .str.strptime(pl.Date, "%m/%d/%Y", strict=False)
    .dt.strftime("%Y-%m")
    .alias("month"),
    pl.col("VIOLATIONS").str.replace_all(",", "").cast(pl.Float64, strict=False).alias("violations"),
).drop_nulls().group_by("month").agg(pl.col("violations").mean().alias("mean_violations")).sort("month")

speed_camera = pl.read_csv("../speed_violations_150000.csv", schema_overrides={"VIOLATIONS": pl.String}).select(
    pl.col("VIOLATION DATE")
    .str.strptime(pl.Date, "%m/%d/%Y", strict=False)
    .dt.strftime("%Y-%m")
    .alias("month"),
    pl.col("VIOLATIONS").str.replace_all(",", "").cast(pl.Float64, strict=False).alias("violations"),
).drop_nulls().group_by("month").agg(pl.col("violations").mean().alias("mean_violations")).sort("month")

fig, ax = plt.subplots(figsize=(12, 5))
ax.plot(red_light["month"], red_light["mean_violations"], label="Red-light cameras")
ax.plot(speed_camera["month"], speed_camera["mean_violations"], label="Speed cameras")
ax.set(title="Average reported violations per record by month", xlabel="Month", ylabel="Mean violations per record")
ax.tick_params(axis="x", labelrotation=45)
ax.set_xticks(ax.get_xticks()[::12])
ax.grid(axis="y", alpha=0.25)
ax.legend()
fig.tight_layout()
plt.show()

for label, monthly_data in [("Red-light", red_light), ("Speed-camera", speed_camera)]:
    print(label, "lowest:", monthly_data.sort("mean_violations").row(0), "highest:", monthly_data.sort("mean_violations", descending=True).row(0))
```

## Information

The speed-camera line is usually higher and changes more sharply from month to month than the red-light-camera line. In the data shown, the speed-camera average ranges from about 12 violations per record in February 2021 to about 99 in July 2021. The red-light-camera average is steadier, ranging from about 3.8 in January 2015 to about 9.5 in June 2023. Some ups and downs are expected because traffic and enforcement can change with the season. However, the July 2021 speed-camera peak surprised me. The average also appears higher from 2021 onward than it was through 2020. A change that large might reflect a real change in traffic or enforcement, but it could also be caused by changes in the records included in the dataset, camera coverage, or how violations were reported. I would not assume the increase is a direct change in how people drive without checking these possibilities.

The graph uses data from the Chicago Data Portal. The dataset describes red-light and speed-camera violations, and it is not clear whether or how Flock cameras are involved in collecting these records. The plot is useful for describing reported violations over time, but it cannot tell us whether the presence of Flock cameras affected those violations.

## Next Steps

Considering what the plot shows, one question I keep coming back to is "Why did the average number of speed-camera violations rise so sharply in 2021, and does this increase have anything to do with flock cameras?" To connect this analysis to the Flock research question, I would need verified flock camera locations and installation dates, plus traffic outcomes for the same places and time periods. One way for further investigation would be to compare areas before and after installation with similar areas that did not receive a Flock camera.
